```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

# 1. Load data
df = pd.read_csv("/content/aug_train.csv")

# 2. Simple Preprocessing: Drop non-numeric columns for now
# (In a real project, you would encode these instead of dropping them!)
numeric_df = df.select_dtypes(include=['number'])

# 3. Define X and y
# Ensure the target is NOT in X
x = numeric_df.drop(["enrollee_id", "city_development_index"], axis=1)
y = numeric_df["city_development_index"]

# 4. Split with correct parameter name: random_state
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

# 5. Train and Evaluate
model = LinearRegression()
model.fit(x_train, y_train)

print("Score:", model.score(x_test, y_test))
```
